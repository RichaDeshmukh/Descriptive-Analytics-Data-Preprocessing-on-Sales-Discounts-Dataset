
# Descriptive Analytics and Data Preprocessing on Sales Discounts Dataset
This notebook demonstrates step-by-step descriptive statistics, visualization, and preprocessing techniques on the provided sales discounts dataset. Every cell offers detailed explanations and commentary.


```

# Import necessary libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Load sales data
sales_df = pd.read_csv('sales_data_with_discounts.csv')
# Display sample rows for context
sales_df.head()
```


Above, we import essential libraries and load the sales data CSV. The `head()` function helps inspect the first few rows, useful for understanding data structure.


```

# Identify numerical and categorical columns
num_cols = sales_df.select_dtypes(include=np.number).columns.tolist()
cat_cols = sales_df.select_dtypes(exclude=np.number).columns.tolist()

print('Numerical columns:', num_cols)
print('Categorical columns:', cat_cols)
```


This cell detects which columns contain numeric vs. categorical data, aiding in tailored analysis and preprocessing steps.


```

# Compute mean, median, mode, and standard deviation
stats_summary = pd.DataFrame({
    'Mean': sales_df[num_cols].mean(),
    'Median': sales_df[num_cols].median(),
    'Mode': sales_df[num_cols].mode().iloc[0],
    'Std Dev': sales_df[num_cols].std()
})
stats_summary
```


Each metric above provides insight:
- **Mean**: Average value
- **Median**: Middle value
- **Mode**: Most frequent value
- **Std Dev**: Spread or variability of data


```

# Plot histograms for each numerical column
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.histplot(sales_df[col], kde=True)
    plt.title(f'Histogram of {col}')
    plt.show()
```


Histograms above display the distribution and skewness of each numerical column in the dataset.


```

# Boxplot for each numerical column
for col in num_cols:
    plt.figure(figsize=(6, 4))
    sns.boxplot(y=sales_df[col])
    plt.title(f'Boxplot of {col}')
    plt.show()
```


Boxplots help visually detect outliers and the spread (interquartile range) for numeric data.


```

# Example categorical visualization: 'Brand' sales count
plt.figure(figsize=(8, 5))
sales_df['Brand'].value_counts().plot(kind='bar')
plt.title('Sales Count by Brand')
plt.ylabel('Number of Sales')
plt.show()
```


Bar charts visualize frequency distributions for categorical columns like 'Brand'.


```

# Standardize (z-score normalization) all numerical columns
sales_df_standardized = sales_df.copy()
sales_df_standardized[num_cols] = (sales_df[num_cols] - sales_df[num_cols].mean()) / sales_df[num_cols].std()
sales_df_standardized.head()
```


Standardizing is important for many machine learning models and comparisons as it brings all numeric features onto the same scale.


```

# One-hot encode all categorical columns
encoded_df = pd.get_dummies(sales_df_standardized, columns=cat_cols, drop_first=True)
encoded_df.head()
```


One-hot encoding lets us convert category data into numeric form suitable for algorithms.
